# Project Title

Cauvery Hydrological Data Analytics: Water Level, Flow & Reservoir Performance

# **Objective(Problem statment)**

1. To Perform exploratory data analysis to identify flow patterns and storage trends across Two-state datasets.

2. To clean, align dates, and transform hourly telemetry, daily discharge, and agricultural demand tables

3. To Evaluate seasonal irrigation demand, supply shortages, and groundwater reliance across delta districts.

4. To Build time-series charts and comparison graphs to visualize reservoir levels, discharge rates, and supply deficits.

5. To Compare upper-basin storage in Karnataka against downstream water availability in Tamil Nadu.

6. To analyse the reservoir release schedules and mitigate crop water deficits.

# Outcome

1. Standardized and integrated 37,408 records across telemetry, discharge, reservoir, and agricultural datasets into a single analytical pipeline.

2. Discovered that upper-basin Karnataka reservoirs (KRS, Kabini) and downstream Tamil Nadu reservoirs (Mettur, Bhavanisagar) averaged ~60% to 61% capacity utilization.

3. Identified a 268,276 MCFT (4.09%) total water deficit across 6 delta districts during the Kuruvai and Samba crop seasons.

4. Mapped flow rates across 66 CWC discharge stations, recording an average downstream flow of 43.37 $\text{m}^3/\text{s}$ in Tamil Nadu.

5. Determined that delta agricultural regions rely on groundwater extraction for 24.44% of their total irrigation needs to make up for canal shortages.

# River Discharge

```python
# Rename Columns of Discharge
df_discharge.rename(columns={'Data Acquisition Time': 'Observation Date',
    'Manual Daily River Water Discharge (m3/sec)': 'Daily Discharge (m3/s)'}, inplace=True)
df_discharge
```

```python
# Save snapshot BEFORE outlier treatment
df_discharge_before_outlier = df_discharge.copy()
```

```python
# Cap extreme values or investigate
df_discharge = df_discharge[df_discharge['Daily Discharge (m3/s)'] <= 5000].copy()
df_discharge
```

```python
# convert to real datetime
df_discharge['Observation Date'] = (
    df_discharge['Observation Date']
    .astype(str)
    .str.replace(r'(\d{1,2})\.(\d{2})$', r'\1:\2', regex=True)
)
df_discharge['Observation Date'] = pd.to_datetime(
    df_discharge['Observation Date'], format='%d-%m-%Y %H:%M', errors='coerce'
)

# Confirm conversion
print("Nulls after conversion:", df_discharge['Observation Date'].isnull().sum())
print("Rows remaining:", len(df_discharge))

# Add Year
df_discharge['Year'] = df_discharge['Observation Date'].dt.year
# Add Month
df_discharge['Month'] = df_discharge['Observation Date'].dt.month_name()
# Add Month Number
df_discharge['Month Number'] = df_discharge['Observation Date'].dt.month
# Extract Quarter
df_discharge['Quarter'] = df_discharge['Observation Date'].dt.quarter
df_discharge
```

```python
# Season-Based Feature Engineering

def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Southwest Monsoon'
    else:
        return 'Northeast Monsoon'

df_discharge['Season'] = df_discharge['Observation Date'].dt.month.map(get_season)
df_discharge
```

```python
# Station-Wise Discharge Category Classification

def classify_discharge(x):
    rank = x.rank(method='average', pct=True)

    return pd.cut(
        rank,
        bins=[0, 0.25, 0.50, 0.75, 1.00],
        labels=['Very Low', 'Low', 'High', 'Very High'],
        include_lowest=True
    )

df_discharge['Discharge Category'] = (
    df_discharge.groupby('Station')['Daily Discharge (m3/s)']
    .transform(classify_discharge)
)
df_discharge
```

```python
# Station-Wise Previous Day Discharge Calculation

df_discharge = df_discharge.sort_values(
    ['Station', 'Observation Date']
)

df_discharge['Previous Day Discharge (m3/s)'] = (
    df_discharge.groupby('Station')['Daily Discharge (m3/s)']
      .shift(1)
)
df_discharge
```

```python
# Station-Wise Discharge Change Calculation

df_discharge['Discharge Change (m3/s)'] = (
    df_discharge['Daily Discharge (m3/s)']
    - df_discharge['Previous Day Discharge (m3/s)']
)
df_discharge
```

```python
# Station-Wise Discharge Percentage Change Calculation

df_discharge['Discharge Change (%)'] = (
    df_discharge['Discharge Change (m3/s)']
    / df_discharge['Previous Day Discharge (m3/s)']
) * 100
df_discharge
# Replace infinite values (from division by zero when Previous Day Discharge = 0) with NaN
df_discharge['Discharge Change (%)'] = df_discharge['Discharge Change (%)'].replace([np.inf, -np.inf], np.nan)

df_discharge
```

```python
# Drop Source file
df_discharge = df_discharge.drop(columns=['Source_File'])
df_discharge
```

# Treat outliers if required

```python
def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower) | (df[column] > upper)]
    print(f"{column} → Outliers: {len(outliers)}")
    return len(outliers)

print("===== OUTLIER CHECK =====")

print("\n1. Water Level:")
find_outliers(df_water_level, 'Water_Level_Meters')
print("Negative values:", (df_water_level['Water_Level_Meters'] < 0).sum())

print("\n2. Discharge:")
find_outliers(df_discharge, 'Daily Discharge (m3/s)')

print("\n3. Reservoir:")
find_outliers(df_reservoirs, 'Storage_Percentage')
find_outliers(df_reservoirs, 'Inflow')
find_outliers(df_reservoirs, 'Outflow')

print("\n4. Agriculture:")
find_outliers(df_agriculture, 'Deficit_Percentage')
find_outliers(df_agriculture, 'Water_Deficit_MCFT')
```

# River Discharge(Transformation)

```python
# Log transform for heavily skewed discharge columns
df_discharge['Daily Discharge (log)'] = np.log1p(df_discharge['Daily Discharge (m3/s)'])
df_discharge['Previous Day Discharge (log)'] = np.log1p(df_discharge['Previous Day Discharge (m3/s)'])

# Discharge Change (m3/s) can go negative, so use a signed log transform instead
df_discharge['Discharge Change (log)'] = np.sign(df_discharge['Discharge Change (m3/s)']) * np.log1p(np.abs(df_discharge['Discharge Change (m3/s)']))
```

# Reservoirs (Transformation)

```python
# Log transform for skewed reservoir columns
df_reservoirs['Storage_MCFT (log)'] = np.sqrt(df_reservoirs['Storage_MCFT'])
df_reservoirs['Inflow (log)'] = np.log1p(df_reservoirs['Inflow'])
df_reservoirs['Outflow (log)'] = np.log1p(df_reservoirs['Outflow'])

# Net_Flow_Cusecs can be negative (outflow > inflow), so use signed log
df_reservoirs['Net_Flow (log)'] = np.sign(df_reservoirs['Net_Flow_Cusecs']) * np.log1p(np.abs(df_reservoirs['Net_Flow_Cusecs']))
```

```python
# Verify new skewness
print("Storage_MCFT (log) skewness:", df_reservoirs['Storage_MCFT (log)'].skew())
print("Inflow (log) skewness:", df_reservoirs['Inflow (log)'].skew())
print("Outflow (log) skewness:", df_reservoirs['Outflow (log)'].skew())
print("Net_Flow (log) skewness:", df_reservoirs['Net_Flow (log)'].skew())
```

```python
!pip install folium -q

import folium

# Cauvery river flow points, in order
river_points = [
    ("Talacauvery (Start)", 12.3833, 75.4833, "Karnataka", 1),
    ("Harangi Dam", 12.49, 75.91, "Karnataka", 2),
    ("Hemavathi Dam", 12.82, 76.05, "Karnataka", 3),
    ("Kabini Dam", 11.97, 76.35, "Karnataka", 4),
    ("KRS Dam", 12.425, 76.572, "Karnataka", 5),
    ("Hogenakkal Entry", 12.12, 77.77, "Tamil Nadu", 6),
    ("Mettur Dam", 11.8, 77.8, "Tamil Nadu", 7),
    ("Bhavani Join", 11.45, 77.68, "Tamil Nadu", 8),
    ("Grand Anicut (Kallanai)", 10.83, 78.82, "Tamil Nadu", 9),
    ("Poompuhar (End)", 11.14, 79.85, "Tamil Nadu", 10),
]

# Center the map roughly on the river's midpoint
center_lat = sum(p[1] for p in river_points) / len(river_points)
center_lon = sum(p[2] for p in river_points) / len(river_points)

m = folium.Map(location=[center_lat, center_lon], zoom_start=7, tiles='OpenStreetMap')

state_colors = {'Karnataka': 'Red', 'Tamil Nadu': 'Yellow'}

# Draw the connecting flow line (in order)
coords = [(p[1], p[2]) for p in river_points]
folium.PolyLine(coords, color='gray', weight=3, opacity=0.8, dash_array='5').add_to(m)

# Add markers with order numbers and labels
for name, lat, lon, state, order in river_points:
    folium.Marker(
        location=[lat, lon],
        popup=f"{order}. {name} ({state})",
        tooltip=f"{order}. {name}",
        icon=folium.DivIcon(html=f"""
            <div style="background-color:{state_colors[state]};
                        color:Black; border-radius:50%;
                        width:26px; height:26px;
                        display:flex; align-items:center; justify-content:center;
                        border:2px solid black; font-weight:bold; font-size:12px;">
                {order}
            </div>""")
    ).add_to(m)

# Add a legend
legend_html = '''
<div style="position: fixed; bottom: 30px; left: 30px; width: 160px; height: 90px;
            background-color: white; z-index:9999; font-size:13px;
            border:2px solid grey; border-radius:6px; padding: 8px;">
<b>State</b><br>
<i style="background:#D97742; width:12px; height:12px; display:inline-block; border-radius:50%;"></i> Karnataka<br>
<i style="background:#2E86AB; width:12px; height:12px; display:inline-block; border-radius:50%;"></i> Tamil Nadu
</div>
'''
m.get_root().html.add_child(folium.Element(legend_html))

m.save('cauvery_flow_map.html')
m
```
